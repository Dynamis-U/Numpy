# Array Properties and Operations

```python
import numpy as np
arr_1d = np.array([1,2,3])
arr_2d = np.array([[10,20,30],
                [40,50,60]])
arr_3d = np.array([[[1,2],[3,4],[5,6],[7,8]]])

print(arr_1d.ndim)
print(arr_2d.ndim)
print(arr_3d.ndim)
```

```python
arr = np.array([[10,20,30],
                [40,50,60]])
print(arr.size)
```

```python
arr_2d = np.array([[1,2,3],
                  [4,5,6]])
print(arr_2d.shape)   # give no. of rows and column
```

#### .dtype -> data type of element

```python
arr = np.array([1,2,3.9,4,5])
print(arr.dtype)
```

#### Datatype conversion of array element
#### array.astype(newtype)

```python
arr = np.array([1.2,2.8,3.9])
print(arr.dtype)
int_arr = arr.astype(int)

print(int_arr)
print(int_arr.dtype)
```

### Mathematical Operation on all Array elements

```python
arr = np.array([10,20,30])

print(arr + 5)
print(arr * 2)
print(arr ** 2)
```

### Aggregation Functions
#### np.sum(array) -> add all
#### np.mean(array) -> calculate average
#### min(array) -> min
#### max(array) -> max
#### np.std(array) -> standard deviation
#### np.var(array) -> variance

```python
arr = np.array([10,20,30,40,50])
print(np.sum(arr))
print(np.mean(arr))
print(np.min(arr))
print(np.max(arr))
print(np.std(arr))
print(np.var(arr))
```
