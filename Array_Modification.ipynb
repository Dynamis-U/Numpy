### Array Modifications
##### np.insert(array, index, value, axis = None)
##### axis = 0 -> row-wise
##### axis = 1 -> column-wise

```python
import numpy as np
arr = np.array([10,20,30,40,50,60])
print(arr)
new_arr = np.insert(arr, 2, 100, axis = 0)
print(new_arr)
```

```python
arr_2d = np.array([[1,2],[3,4]])
print(arr_2d)
new_arr_2d = np.insert(arr_2d,1, [5,6], axis = 1)
print(new_arr_2d)
```

 #### Append
 

```python
arr = np.array([10,20,30])
new_arr = np.append(arr, [40,50,60])
print(new_arr)
```

#### Concatenate
##### np.concatenate((array1, array2),axis)
##### axis = 0 -> vertical stacking
##### axis = 1 -> horizontal stacking

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
new_arr = np.concatenate((arr1,arr2))
print(new_arr)
```

#### np.delete(array, index, axis = None)
##### flatten array

```python
arr = np.array([10,20,30,40,50,60])
print(arr)
new_arr = np.delete(arr,0)
print(new_arr)
```

```python
arr_2d = np.array([[1,2,3],[4,5,6]])
new_arr_2d = np.delete(arr_2d,0,axis = 1)
print(new_arr_2d)
```

### Stacking and Spliting
##### vstack() row wise
##### hstack() column wise

```python
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
print(np.hstack((arr1,arr2)))
```

#### np.split() -> equal parts
#### np.hsplit()
#### np.vsplit() -> work with more than two dimension

```python
arr = np.array([10,20,30,40,50,60])
print(arr)
# print(np.split(arr, 2))
print(np.hsplit(arr,2))
```
