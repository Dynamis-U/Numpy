### Comparing List with Numpy Array

```python
python_list = [1,2,3,4,5]
print(python_list)

import numpy as np   # importing here but this np can be used in further below cell too..
numpy_array = np.array([1,2,3,4,5])
print(numpy_array)
```

```python
ar_1d = np.array([10,20,30,40,50])

print(ar_1d)
```

### Creating 2D Array :-

```python
arr_2D = np.array([[1,2,3],  #also this is a matrix
                  [4,5,6],
                  [7,8,9]])
print(arr_2D)
```

### Array creating with default values



```python
zeroes_array = np.zeros(3)
print(zeroes_array)

zeroes_array = np.ones((2,3))
print(zeroes_array)

filled_Array = np.full((2,2),7)
print(filled_Array)
```

### Creating sequence of number in numpy 
#### arange()                  arange(start,stop,step)

```python
arr = np.arange(1,10,2)
print(arr)
```

### Creating identity matrices
#### eye(size)

```python
identity_matrix = np.eye(5)
print(identity_matrix)
```
