#### np.isnan(array) -> detect missing values
#### np.nan_to_num(array, nan = value(default = 0)) -> replace missing value with number
#### np.isinf(array) -> detect infinity values

```python
import numpy as np

arr = np.array([1,2,np.nan,4,np.nan,6])

print(np.isnan(arr))

print(np.nan == np.nan) # we can not compare np.nan values directly
```

```python
cleaned_arr = np.nan_to_num(arr, nan = 100)
print(cleaned_arr)
```

```python
arr = np.array([1, 2, np.inf, 4, -np.inf, 6])
print(np.isinf(arr))
```

```python
cleaned_arr = np.nan_to_num(arr, posinf = 1000, neginf = -1000)
print(cleaned_arr)
```
